```python
import pandas as pd
import ast

resume_df = pd.read_csv("data/raw/resumes/Resume.csv")
jobs_df = pd.read_csv("data/raw/job_descriptions/job_title_des.csv")

from app.ml.preprocessing import convert_skills_to_list

resume_df["clean_skills"] = resume_df["skills"].apply(convert_skills_to_list)
```

```python
def convert(skill_string):
    try:
        return ast.literal_eval(skill_string)
    except:
        return []
```

```python
resume_df["clean_skills"] = resume_df["skills"].apply(convert)
```

```python
all_skills = []

for skills in resume_df["clean_skills"]:
    for skill in skills:
        all_skills.append(skill.strip().lower())
```

```python
unique_skills = sorted(set(all_skills))

print(len(unique_skills))
```
